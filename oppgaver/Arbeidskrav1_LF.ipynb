## Oppgave 1 – Din første Python-kode

```python
# Lagre navn og alder i variabler
navn = "Markus"
alder = 29

# Skriv ut verdiene
print("Hei! Jeg heter", navn, "og er", alder, "år gammel.")
```

# Oppgave 2 – Regneoperasjoner

```python
# Lager to variabler
x = 10
y = 5

# Regneoperasjoner
addisjon = x + y
subtraksjon = x - y
multiplikasjon = x * y
divisjon = x / y

# Skriver ut resultatene
print("Addisjon:", addisjon)
print("Subtraksjon:", subtraksjon)
print("Multiplikasjon:", multiplikasjon)
print("Divisjon:", divisjon)
```

# Oppgave 3 - Valutakalkulator

```python
# Spør brukeren hvor mange NOK de har
nok = float(input("Hvor mange kroner har du? "))

# Valutakurs
kurs = 11.50

# Konverterer NOK til Euro
euro = round(nok / kurs, 2) # runder av til 2 desimaler

# Skriver ut resultatet
print("Du har", nok, "norske kroner, som tilsvarer",euro, "Euro.")
print(f"Du har {nok} norske kroner, som tilsvarer {euro} Euro.") # her bruker jeg en f-string. Ser du forskjellen på linja over?
```

# Oppgave 4 – Enkel rentevekst

```python
# Startkapital og rente
startkapital = 10000
rente = 0.05

# Beregner kapital etter ett år
ny_kapital = startkapital * (1 + rente)

# Skriver ut resultatet
print("Etter ett år har du", ny_kapital, "kroner.")
```

# Oppgave 5 – Datatyper

```python
# Lager variabler med forskjellige datatyper
heltall = 10
desimaltall = 3.14
tekst = "Hei!"
sann_usann = True

# Skriver ut datatypene
print(type(heltall))
print(type(desimaltall))
print(type(tekst))
print(type(sann_usann))
```
