# Oppgave 2 – Funksjon med argument
Lag en funksjon som heter `kvadrat(x)`, som har som mål å kvadrere ett tall $x$.

````python
# definerer funksjonen kvadrat, med x som parameter
def kvadrat(x):
    return x**2 # returnerer x kvadrert

test_tall1 = 2
test_tall2 = 4
test_tall3 = 7
# Tester funksjonen vår
print(f'Vi gir funksjonen vår {test_tall1} returnerer funksjonen kvadrat(x) {kvadrat(test_tall1)}')
print(f'Vi gir funksjonen vår {test_tall2} returnerer funksjonen kvadrat(x) {kvadrat(test_tall2)}')
print(f'Vi gir funksjonen vår {test_tall3} returnerer funksjonen kvadrat(x) {kvadrat(test_tall3)}')
````

# Oppgave 3 – Valutakonvertering
Lager en funksjon som heter `til_euro(kroner)`


````python
# Vi definerer funksjonen, som tar inn norske kroner og skal gi ut euro
def til_euro(kroner):
    valutakurs = 11.50          # valutakurs
    euro = kroner/valutakurs    # omregning
    return round(euro, 2)       # runder av til 2 desimaler

# Eksempelverdier
test_1 = 100  # 100 norske kroner
test_2 = 1500 # 1500 norske kroner

# Kaller på funksjonen
euro_1 = til_euro(test_1)
euro_2 = til_euro(test_2)

# print-setninger ved hjelp av f-setninger
print(f'{test_1} kr er det samme som {euro_1} €')
print(f'{test_2} kr er det samme som {euro_2} €')
````

# Oppgave 4 – Beregne rentevekst
Her skal vi lage en funksjon som skal hete `rentevekst(startkapital, rente)`.

Husk at `startkapital` og `rente` er **parametere**, altså *placeholders* som skal brukes av faktiske verdier når vi *kaller* på funksjonen. Når vi gir *paramterne* våre faktisk verdier, er disse **argument**!

````python
def rentevekst(startkapital, rente):
    ny_kapital = startkapital * (1 + rente)
    return round(ny_kapital, 2) # runder av til to desimal

print(f'Hvis du starter med 1000 kr og har en rente på 5% vil du etter ett år ha {rentevekst(1000, 0.05)} kr')
print(f'Hvis du starter med 5000 kr og har en rente på 3% vil du etter ett år ha {rentevekst(5000, 0.03)} kr')
````

# Oppgave 5 - Gjennomsnitt
Her skal jeg lage en funksjon som heter `gjennomsnitt(a, b, c)`.

````python
def gjennomsnitt(a, b, c):
    summen = a + b + c
    gjennomsnittet = summen / 3
    return gjennomsnittet

print(f'Gjennomsnittet av 5, 10 og 15 er {gjennomsnitt(5, 10, 15)}')
print(f'Gjennomsnittet av 20, 30 og 40 er {gjennomsnitt(20, 30, 40)}')
````

# Oppgave 6 - Temperaturkonvertering
Lager en funksjon som konverterer celcius om til fahrenheit.

````python
def celsius_til_fahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return round(fahrenheit, 2)

print(f'20 grader Celsius er {celsius_til_fahrenheit(20)} grader Fahrenheit')
````

````python
# Lager en funksjon som regner ut arealet av en sirkel
import math
def areal_sirkel(radius):
    areal = math.pi * (radius ** 2)
    return round(areal, 2)

print(f'Arealet av en sirkel med radius 5 er {areal_sirkel(5)}')
````
