## Import av Pandas

Vi importerer Pandas én gang i starten av notebooken.


````python
# Importerer Pandas med standardforkortelsen (aliaset) pd
import pandas as pd
````

## Oppgave 1 – Lage en DataFrame

Lag en DataFrame med informasjon om navn, alder og studie.

Programmet skal:

- Printe DataFrame-en
- Printe de første radene med `head()`
- Printe dimensjonen med `shape`


````python
# Lager et dictionary med dataene
studentdata = {
    "Navn": ["Anna", "Ole", "Kari"],
    "Alder": [21, 22, 20],
    "Studie": ["Økonomi", "Finans", "Regnskap"]
}

# Gjør dictionary-et om til en DataFrame
studenter = pd.DataFrame(studentdata)

# Printer hele DataFrame-en
print("Hele DataFrame-en:")
print(studenter)

# head() viser som standard de første fem radene
print("\nFørste rader:")
print(studenter.head())

# shape gir (antall rader, antall kolonner)
print("\nDimensjon:")
print(studenter.shape)
````

## Oppgave 2 – Velge kolonner

Bruk DataFrame-en fra oppgave 1.

Programmet skal:

- Printe kolonnen `Navn`
- Printe kolonnen `Alder`
- Finne gjennomsnittsalderen


````python
# Velger og printer kolonnen Navn
print("Navn:")
print(studenter["Navn"])

# Velger og printer kolonnen Alder
print("\nAlder:")
print(studenter["Alder"])

# mean() beregner gjennomsnittet
gjennomsnittsalder = studenter["Alder"].mean()

print("\nGjennomsnittsalder:", gjennomsnittsalder)
````

## Oppgave 3 – Filtrering

Filtrer ut alle studenter som er eldre enn 20 år.


````python
# Lager en boolsk betingelse: Alder må være større enn 20
eldre_enn_20 = studenter[studenter["Alder"] > 20]

print(eldre_enn_20)
````

## Oppgave 4 – Lage ny kolonne

Lag en ny kolonne som heter `Alder_om_5_aar`.

Kolonnen skal inneholde alderen til studentene om fem år.


````python
# Pandas utfører + 5 på alle radene i kolonnen samtidig
studenter["Alder_om_5_aar"] = studenter["Alder"] + 5

print(studenter)
````

## Oppgave 5 – Sortering

Sorter DataFrame-en etter alder i stigende rekkefølge.


````python
# sort_values() sorterer radene etter verdiene i valgt kolonne
studenter_sortert = studenter.sort_values(by="Alder")

print(studenter_sortert)
````

## Oppgave 6 – Lese inn CSV-fil

Last ned én CSV-fil fra `data`-mappen på kurssiden og legg filen i samme mappe som notebooken.

Endre verdien til `filnavn` under slik at den samsvarer med filen du har lastet ned.

Eksempel:

```python
filnavn = "min_datafil.csv"
```


````python
# BYTT UT med navnet på CSV-filen du har valgt fra kurssiden
filnavn = "filnavn.csv"

# Leser CSV-filen inn som en DataFrame
data = pd.read_csv(filnavn)

# Viser de første fem radene
print("Første fem rader:")
print(data.head())

# Printer kolonnenavnene
print("\nKolonnenavn:")
print(data.columns)

# shape gir antall rader og kolonner
print("\nAntall rader og kolonner:")
print(data.shape)

print(f"Datasettet har {data.shape[0]} rader og {data.shape[1]} kolonner.")
````

## Oppgave 7 – Enkle analyser

Vi skal finne:

- Gjennomsnittet i en numerisk kolonne
- Største verdi
- Minste verdi

Siden studentene kan velge forskjellige datasett, finner vi først hvilke kolonner som er numeriske.


````python
# Finner alle numeriske kolonner i datasettet
numeriske_kolonner = data.select_dtypes(include="number").columns

print("Numeriske kolonner:")
print(list(numeriske_kolonner))
````

````python
# Velger den første numeriske kolonnen i datasettet
valgt_kolonne = numeriske_kolonner[0]

print("Valgt kolonne:", valgt_kolonne)

# Beregner enkle statistiske mål
gjennomsnitt = data[valgt_kolonne].mean()
storste_verdi = data[valgt_kolonne].max()
minste_verdi = data[valgt_kolonne].min()

print("Gjennomsnitt:", gjennomsnitt)
print("Største verdi:", storste_verdi)
print("Minste verdi:", minste_verdi)
````

Du kan også velge en bestemt numerisk kolonne selv, for eksempel:

```python
valgt_kolonne = "Pris"
```

dersom datasettet inneholder en kolonne med dette navnet.


## Oppgave 8 – Manglende verdier

Bruk `isnull().sum()` for å finne antall manglende verdier i hver kolonne.


````python
# isnull() markerer manglende verdier.
# sum() teller hvor mange manglende verdier hver kolonne har.
manglende_verdier = data.isnull().sum()

print("Manglende verdier per kolonne:")
print(manglende_verdier)
````

Vi kan i tillegg finne totalt antall manglende verdier i hele datasettet.


````python
totalt_manglende = data.isnull().sum().sum()

print("Totalt antall manglende verdier:", totalt_manglende)
````

## Oppgave 9 – Utfordring

Lag en ny kolonne basert på en beregning mellom to eksisterende kolonner.

Hvilke kolonner som passer, avhenger av datasettet du valgte i oppgave 6.

Under vises først en generell løsning. Vi bruker de to første numeriske kolonnene og beregner forholdet mellom dem.


````python
# Henter navnene på de to første numeriske kolonnene
kolonne_1 = numeriske_kolonner[0]
kolonne_2 = numeriske_kolonner[1]

# Lager en ny kolonne som viser forholdet mellom kolonne 1 og kolonne 2.
# Dette viser hvor mye vi har av kolonne 1 per enhet av kolonne 2.
data["Forhold"] = data[kolonne_1] / data[kolonne_2]

print(data[[kolonne_1, kolonne_2, "Forhold"]].head())
````

### Eksempel på en mer meningsfull beregning

Dersom datasettet for eksempel inneholder kolonnen `Pris`, kan vi lage pris inkludert 25 % moms:

```python
data["Pris_med_moms"] = data["Pris"] * 1.25
```

Hvis datasettet inneholder `Poengsum` og `Timer`, kan vi for eksempel beregne:

```python
data["Poeng_per_time"] = data["Poengsum"] / data["Timer"]
```

Det beste er å velge to kolonner der beregningen har en naturlig tolkning.
