## Oppgave 1 – Symboler og forenkling av uttrykk

Vi skal forenkle uttrykket

$$
f(x)=\frac{x^2-4}{x-2}.
$$

Først definerer vi $x$ som et symbol. Deretter skriver vi brøken i Python og bruker `sp.simplify()` eller `sp.cancel()` til å forenkle den.

```python
# Definerer x som et matematisk symbol
x = sp.symbols("x")

# Uttrykket f(x) fra oppgaven
f = (x**2 - 4) / (x - 2)

# Forenkler den algebraiske brøken
forenklet = sp.simplify(f)

print("Opprinnelig uttrykk f(x):", f)
print("Forenklet uttrykk:", forenklet)
```

**Forklaring:** Telleren er en differanse mellom to kvadrater:

$$
\frac{x^2-4}{x-2}=\frac{(x-2)(x+2)}{x-2}=x+2, \qquad x\neq 2.
$$

Dermed får vi $f(x)=x+2$ for alle $x$ der det opprinnelige uttrykket er definert.

**Viktig presisering:** Det opprinnelige uttrykket er **ikke definert ved $x=2$**, selv om $x+2$ isolert sett er definert der. Ved forkorting må vi beholde denne begrensningen.

**Tips:** Potenser skrives med `**` i Python, for eksempel `x**2` for $x^2$. `sp.cancel(f)` er et alternativ til `sp.simplify(f)` for rasjonale uttrykk.

## Oppgave 2 – Enkel derivasjon

Bedriften har kostnadsfunksjonen

$$
K(x)=2x^3-15x^2+50x+100,
$$

der $x$ er antall produserte enheter.

**Grensekostnaden** er den deriverte av kostnadsfunksjonen:

$$
K'(x)=6x^2-30x+50.
$$

Vi beregner deretter $K'(10)$ ved å sette inn $x=10$.

```python
# Kostnadsfunksjonen fra oppgaven
K = 2*x**3 - 15*x**2 + 50*x + 100

# Deriverer K(x) med hensyn på x
grensekostnad = sp.diff(K, x)

# Setter inn produksjonsnivået x = 10
grensekostnad_10 = grensekostnad.subs(x, 10)

print("Kostnadsfunksjon K(x):", K)
print("Grensekostnad K'(x):", grensekostnad)
print("Grensekostnad ved x = 10:", grensekostnad_10, "kr per enhet")
```

**Utregning og tolkning:**

$$
K'(10)=6\cdot10^2-30\cdot10+50=600-300+50=350.
$$

Grensekostnaden ved produksjon på 10 enheter er **350 kr per enhet**. Det betyr at kostnadene øker med om lag 350 kr for en liten økning på én enhet når produksjonen ligger rundt 10 enheter. Den deriverte er en *marginal* endringsrate, ikke nødvendigvis den eksakte kostnadsøkningen fra 10 til 11 enheter.

**Tips:** `sp.diff(K, x)` finner den deriverte symbolsk. `.subs(x, 10)` setter inn en verdi i uttrykket.

## Oppgave 3 – Derivasjon med kjerneregelen

Vi skal derivere

$$
g(x)=(3x^2+2x)^4.
$$

Her er $u(x)=3x^2+2x$ den indre funksjonen, og den ytre funksjonen er $u^4$.

Kjerneregelen gir

$$
g'(x)=4(3x^2+2x)^3\,(6x+2).
$$

SymPy anvender kjerneregelen automatisk.

```python
# Funksjonen g(x) fra oppgaven
g = (3*x**2 + 2*x)**4

# Deriverer g(x) med hensyn på x
g_derivert = sp.diff(g, x)

# factor() gir en alternativ, faktorisert form
print("Funksjonen g(x):", g)
print("Den deriverte g'(x):", g_derivert)
print("Faktorisert svar:", sp.factor(g_derivert))
```

**Forklaring:** Først deriverer vi den ytre potensen: $4u^3$. Deretter multipliserer vi med den deriverte av det indre uttrykket, $u'(x)=6x+2$.

Vi får dermed

$$
g'(x)=4(6x+2)(3x^2+2x)^3.
$$

**Tips:** SymPy kan skrive resultatet på en annen, men matematisk ekvivalent måte. `sp.factor()` kan brukes til å faktorisere et uttrykk, mens `sp.expand()` utvider parentesene.

## Oppgave 4 – Økonomisk likevekt: Tilbud og etterspørsel

Vi har et marked med prisfunksjonene

$$
P_d(Q)=200-4Q, \qquad P_s(Q)=50+2Q.
$$

$P_d$ er etterspørselsprisen (betalingsvillighet), og $P_s$ er tilbudsprisen (produsentenes priskrav).

I markedslikevekt er prisene like:

$$
200-4Q=50+2Q.
$$

Vi bruker SymPy til først å løse for likevektsmengden $Q^*$ og deretter beregne likevektsprisen $P^*$.

```python
# Definerer mengden Q som symbol
Q = sp.symbols("Q")

# Etterspørsels- og tilbudsfunksjonene fra oppgaven
Pd = 200 - 4*Q  # Kundenes betalingsvillighet
Ps = 50 + 2*Q   # Produsentenes priskrav

# Setter etterspørsel lik tilbud
likevektsligning = sp.Eq(Pd, Ps)

# Løser ligningen med hensyn på Q
losninger = sp.solve(likevektsligning, Q)
likevektsmengde = losninger[0]

# Setter likevektsmengden inn i etterspørselsfunksjonen
likevektspris = Pd.subs(Q, likevektsmengde)

print("Likevektsligning:", likevektsligning)
print("Løsninger for Q:", losninger)
print("Likevektsmengde Q*:", likevektsmengde, "enheter")
print("Likevektspris P*:", likevektspris, "kr")
```

**Løsning for hånd:**

$$
200-4Q=50+2Q
\quad\Longleftrightarrow\quad
150=6Q
\quad\Longleftrightarrow\quad
Q^*=25.
$$

Vi setter deretter $Q^*=25$ inn i én av prisfunksjonene:

$$
P^*=200-4\cdot25=100.
$$

**Svar:** Likevektsmengden er **25 enheter**, og likevektsprisen er **100 kr**.

Som ekstra kontroll kan vi sette $Q=25$ inn i begge prisfunksjonene. De skal gi samme pris.

```python
# Kontrollerer at tilbud og etterspørsel gir samme pris
pris_fra_ettersporsel = Pd.subs(Q, likevektsmengde)
pris_fra_tilbud = Ps.subs(Q, likevektsmengde)

print("Pris fra etterspørsel:", pris_fra_ettersporsel, "kr")
print("Pris fra tilbud:", pris_fra_tilbud, "kr")
print("Er prisene like?", pris_fra_ettersporsel == pris_fra_tilbud)

assert pris_fra_ettersporsel == pris_fra_tilbud
```

**Tolkning:** Ved en mengde på 25 enheter er både kundenes betalingsvillighet og produsentenes priskrav 100 kr. Dette er markedslikevekten i modellen.

**Tips:** `sp.Eq(Pd, Ps)` lager ligningen. `sp.solve(..., Q)` finner verdien(e) av $Q$ som oppfyller den. `.subs()` brukes deretter til å finne tilhørende pris.
