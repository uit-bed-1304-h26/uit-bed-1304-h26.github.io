## Oppgave 1 – Kvantumsrabatt

En bedrift selger en vare for 200 kroner per stykk.

Rabatten er:

- 0 % ved kjøp av færre enn 50 enheter
- 10 % ved kjøp mellom 50 og 99 enheter
- 20 % ved kjøp av 100 enheter eller mer

Programmet skal beregne totalprisen etter rabatt.


````python
# Pris per vare
pris_per_enhet = 200

# Eksempel: kunden ønsker å kjøpe 75 varer
antall_varer = 75

# Bestemmer rabatten ved hjelp av if, elif og else
if antall_varer < 50:
    rabatt = 0
elif antall_varer < 100:
    rabatt = 0.10
else:
    rabatt = 0.20

# Pris før rabatt
pris_for_rabatt = antall_varer * pris_per_enhet

# Beregner totalpris etter rabatt
totalpris = pris_for_rabatt * (1 - rabatt)

print(f"Antall varer: {antall_varer}")
print(f"Rabatt: {rabatt:.0%}")
print(f"Totalpris: {totalpris:.2f} kr")
````

## Oppgave 2 – Rentes rente over tid

Du investerer 50 000 kroner i et indeksfond med forventet årlig avkastning på 7 %.

Bruk en `for`-løkke til å beregne saldoen for hvert av de neste 10 årene.


````python
# Startbeløp og årlig avkastning
saldo = 50000
rente = 0.07

# range(1, 11) gir år 1 til og med år 10
for aar in range(1, 11):
    # Saldoen øker med 7 % hvert år
    saldo = saldo * (1 + rente)

    print(f"År {aar}: {saldo:.2f} kr")
````

Her oppdateres `saldo` én gang for hver gjennomkjøring av løkken. Dermed blir avkastningen året etter beregnet av både det opprinnelige beløpet og tidligere avkastning.


## Oppgave 3 – Investeringsmål

Du starter med 100 000 kroner og ønsker å komme over 500 000 kroner.

Hvert år:

1. Saldoen øker med 4 % rente.
2. Du setter inn 30 000 kroner ved slutten av året.

Bruk en `while`-løkke til å finne hvor mange år det tar.


````python
# Startverdier
saldo = 100000
maal = 500000
rente = 0.04
aarlig_innskudd = 30000
aar = 0

# Løkken fortsetter så lenge saldoen er under målet
while saldo < maal:
    # Først legges årets rente til
    saldo = saldo * (1 + rente)

    # Deretter settes årets sparebeløp inn
    saldo = saldo + aarlig_innskudd

    # Ett nytt år har gått
    aar = aar + 1

    print(f"År {aar}: {saldo:.2f} kr")

print(f"\nDet tar {aar} år før saldoen passerer {maal} kr.")
````

`while`-løkken stopper automatisk når betingelsen `saldo < maal` ikke lenger er sann.


## Oppgave 4 – Transaksjonsanalyse

Vi har følgende kontantstrøm:

```python
kontantstrom = [15000, -3500, -1200, 8000, -450, -2000, 5000]
```

Positive tall er inntekter og negative tall er utgifter.

Bruk en `for`-løkke og `if / else` til å finne:

- Summen av alle inntekter
- Summen av alle utgifter


````python
# Liste med bedriftens transaksjoner
kontantstrom = [15000, -3500, -1200, 8000, -450, -2000, 5000]

# Starter summeringen på 0
sum_inntekter = 0
sum_utgifter = 0

# Går gjennom én transaksjon om gangen
for transaksjon in kontantstrom:

    if transaksjon >= 0:
        # Positive verdier legges til inntektene
        sum_inntekter = sum_inntekter + transaksjon
    else:
        # Negative verdier legges til utgiftene
        sum_utgifter = sum_utgifter + transaksjon

print(f"Sum inntekter: {sum_inntekter} kr")
print(f"Sum utgifter: {sum_utgifter} kr")
````

Siden utgiftene er lagret som negative tall, blir `sum_utgifter` også negativ.

Hvis vi ønsker å vise utgiftene som et positivt beløp, kan vi bruke `abs()`:


````python
print(f"Utgifter som positivt beløp: {abs(sum_utgifter)} kr")
````
