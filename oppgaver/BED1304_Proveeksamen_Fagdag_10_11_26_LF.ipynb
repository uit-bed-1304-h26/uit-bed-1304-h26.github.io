### Oppgave 2.4 – Pseudokode og if/elif/else (5 p)
En nettbutikk skal beregne frakt for en ordre. Reglene er:

- Ordrer på 1000 kr eller mer får gratis frakt (0 kr).
- Ordrer fra og med 500 kr (men under 1000 kr) betaler 49 kr i frakt.
- Alle andre ordrer betaler 99 kr i frakt.

a) Skriv pseudokode for en funksjon `beregn_frakt(belop)`. (2 p)
b) Skriv funksjonen i Python. Den skal **returnere** frakten. (2 p)
c) Hva hadde gått galt hvis du sjekket `belop >= 500` *før* `belop >= 1000`? (1 p)

✅ **a) Pseudokode**
- Ta inn beløp
- HVIS beløp ≥ 1000: returner 0
- ELLERS HVIS beløp ≥ 500: returner 49
- ELLERS: returner 99

**c)** En ordre på f.eks. 1500 kr ville stoppet på `belop >= 500` og fått 49 kr i frakt. Python stopper ved første sanne betingelse, så den strengeste må stå først.

**Typisk feil:** `print` i stedet for `return`, eller `>` i stedet for `>=` (da får en ordre på nøyaktig 1000 kr frakt).

````python
def beregn_frakt(belop):
    if belop >= 1000:
        return 0
    elif belop >= 500:
        return 49
    else:
        return 99

print(beregn_frakt(1500))  # 0
print(beregn_frakt(1000))  # 0
print(beregn_frakt(700))   # 49
print(beregn_frakt(200))   # 99
````

### Oppgave 2.6 – SymPy: markedslikevekt (5 p)
Invers tilbud og etterspørsel er gitt ved

- Tilbud: $P = 10 + 2Q$
- Etterspørsel: $P = 100 - Q$

a) Skriv SymPy-koden som finner likevektsmengden $Q^*$. Bruk `sp.Symbol`, `sp.Eq` og `sp.solve`. (3 p)
b) Bruk `.subs()` til å finne likevektsprisen $P^*$. (1 p)
c) Hva blir $Q^*$ og $P^*$? (1 p)

✅ **c)** $Q^* = 30$ og $P^* = 70$. Sjekk for hånd: $10 + 2Q = 100 - Q \Rightarrow 3Q = 90 \Rightarrow Q = 30$.

**Typisk feil:**
- Glemmer `[0]` etter `sp.solve`, så `subs` får en liste i stedet for et tall.
- Skriver `^` i stedet for `**` (ikke relevant her, men vanlig i SymPy-oppgaver).

````python
Q = sp.Symbol('Q')
tilbud = 10 + 2*Q
ettersporsel = 100 - Q

# a)
Q_stjerne = sp.solve(sp.Eq(tilbud, ettersporsel), Q)[0]

# b)
P_stjerne = ettersporsel.subs(Q, Q_stjerne)

print(f"Q* = {Q_stjerne}, P* = {P_stjerne}")
````

### Oppgave 2.7 – Feilsøking (5 p)
Koden skal regne ut lønn etter skatt for 37,5 timer à 250 kr med 25 % skatt. Den inneholder **fem feil**.

```python
def lonn_etter_skatt(timer, timelonn, skattesats)
    brutto = timer * timelonn
    if brutto = 0:
        print("Ingen lønn")
        netto = 0
    else:
        netto = brutto * (1 - skattesats)
  return netto

resultat = lonn_etter_skatt(37.5, 250, 25)
print("Utbetalt: " + resultat + " kr")
```

Finn de fem feilene. For hver feil: si om det er en syntaksfeil, en kjøretidsfeil (f.eks. `TypeError`) eller en logisk feil, og vis hvordan den rettes. (1 p per feil)

✅
1. Mangler `:` etter `def`-linjen – **syntaksfeil**.
2. `if brutto = 0:` skal være `==` – **syntaksfeil**.
3. `return` har feil innrykk – **syntaksfeil** (`IndentationError`).
4. `skattesats` sendes inn som `25` i stedet for `0.25` – **logisk feil** (gir netto = −225 000 kr, uten feilmelding).
5. `"Utbetalt: " + resultat` blander tekst og tall – **kjøretidsfeil** (`TypeError`). Bruk f-streng.

Riktig svar: `Utbetalt: 7031.25 kr`.

**Poeng til seminar:** Den logiske feilen er farligst, fordi Python ikke klager. Koble til Pólyas steg 4: «Tror du på svaret?»

````python
def lonn_etter_skatt(timer, timelonn, skattesats):
    brutto = timer * timelonn
    if brutto == 0:
        print("Ingen lønn")
        netto = 0
    else:
        netto = brutto * (1 - skattesats)
    return netto

resultat = lonn_etter_skatt(37.5, 250, 0.25)
print(f"Utbetalt: {resultat:.2f} kr")
````

### Oppgave 3.1 – Nedbetaling av studielån (15 p)
Du har et studielån og vil vite hvor mange år det tar å betale det ned.

Hvert år skjer følgende:
1. Det legges til renter på lånet: `laan = laan * (1 + rente)`
2. Du betaler et fast årlig beløp, som trekkes fra lånet.

Dette gjentas så lenge det er noe igjen av lånet (`laan > 0`).

**Oppgave:**

a) Skriv pseudokode for programmet. (3 p)

b) Lag en funksjon `nedbetaling(laan, rente, aarlig_betaling)` som bruker en `while`-løkke og **returnerer** antall år det tar å betale ned lånet. (8 p)

c) Kall funksjonen med et lån på 200 000 kr, 5 % rente og 25 000 kr i året. Skriv ut svaret med en f-streng, f.eks. `Lånet er nedbetalt etter X år.` (4 p)

✅ **a) Pseudokode**
- Ta inn lån, rente og årlig betaling
- Sett år = 0
- SÅ LENGE lån > 0:
  - legg til renter: lån = lån × (1 + rente)
  - trekk fra betalingen: lån = lån − årlig betaling
  - øk år med 1
- Returner år

**Vurdering (forslag):** `def` med riktige parametere (2 p), løkke med riktig betingelse (2 p), riktig oppdatering av lån og teller (2 p), `return` (2 p), riktig kall og utskrift med f-streng (4 p), pseudokode (3 p).

**Typisk feil:**
- Glemmer `aar += 1` → uendelig løkke.
- `print` inne i funksjonen i stedet for `return`.
- Bruker `rente = 5` i stedet for `0.05`.

**Utvidelse for de raske:** Hva skjer hvis den årlige betalingen er 10 000 kr? Legg inn en `if`-sjekk før løkken som stopper funksjonen når betalingen ikke dekker rentene.

````python
def nedbetaling(laan, rente, aarlig_betaling):
    aar = 0

    # Gjenta så lenge det er noe igjen av lånet
    while laan > 0:
        laan = laan * (1 + rente)      # legg til renter
        laan = laan - aarlig_betaling  # trekk fra betalingen
        aar += 1                       # ett år har gått

    return aar


# c)
antall_aar = nedbetaling(200000, 0.05, 25000)
print(f"Lånet er nedbetalt etter {antall_aar} år.")
````

```python
data = {
    "Ticker":      ["EQNR", "DNB", "MOWI", "YAR", "TEL", "NHY"],
```

**Oppgave:**

b) Lag en ny kolonne `Markedsverdi` = Antall × Dagens kurs. (3 p)

d) Regn ut og skriv ut den samlede markedsverdien til porteføljen. (2 p)

e) Lag et søylediagram (`plt.bar`) med ticker på x-aksen og markedsverdi på y-aksen. Figuren skal ha tittel og aksetitler. (4 p)

````python
# a)
data = {
    "Ticker":      ["EQNR", "DNB", "MOWI", "YAR", "TEL", "NHY"],
    "Antall":      [100, 150, 200, 40, 120, 300],
    "Kjopskurs":   [280.0, 190.0, 205.0, 380.0, 140.0, 70.0],
    "Dagens_kurs": [265.0, 225.0, 195.0, 410.0, np.nan, 78.0],
}
df = pd.DataFrame(data)
df = df.dropna()

# b)
df["Markedsverdi"] = df["Antall"] * df["Dagens_kurs"]
print(df)

# c)
tap = df[df["Dagens_kurs"] < df["Kjopskurs"]]
print("\nAksjer med tap:", list(tap["Ticker"]))

# d)
total = df["Markedsverdi"].sum()
print(f"\nSamlet markedsverdi: {total:.2f} kr")

# e)
plt.bar(df["Ticker"], df["Markedsverdi"])
plt.title("Markedsverdi per aksje i porteføljen")
plt.xlabel("Aksje (ticker)")
plt.ylabel("Markedsverdi (NOK)")
plt.show()
````
